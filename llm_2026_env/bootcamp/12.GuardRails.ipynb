### Guard Rails
It can be of 2 types:
1. Deterministic Approach: NLP Techniques (Regx)
2. LLM Based Approac: using LLM, Semantic routing, Middleware(langchain)

Langchain GuardRails: like PII Middleware, Human-in-the-loop,Before Agent Call, After Agent Call


```python
#load the API

import os 
from dotenv import find_dotenv, load_dotenv

load_dotenv(find_dotenv(usecwd=True), override=True)
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["TAVILY_API_KEY"] = os.getenv("TAVILY_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
print("API Key Loaded:", os.getenv("GROQ_API_KEY")[:10] if os.getenv("GROQ_API_KEY") else "FAILED TO LOAD")

#Test the API KEY
from langchain_groq import ChatGroq
from langchain_openai import ChatOpenAI #alternative 

llm_model = ChatGroq(model ="llama-3.3-70b-versatile" )
print(llm_model.invoke("hello, tell me about yourself in few words").content)
```

```python
#1. Redact email addresses
from langchain.agents import create_agent
from langchain.agents.middleware import PIIMiddleware

agent = create_agent(
    model = llm_model,
    middleware=[
        PIIMiddleware(
            "email",
            strategy="redact"
        ),
    ],
)

response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "My email is john.smith@gmail.com"
        }
    ]
})

print(response["messages"][-1].content)
```

```python
#### Strategy: Mask

from langchain.agents import create_agent
from langchain.agents.middleware import PIIMiddleware

agent = create_agent(
    model = llm_model,
    middleware=[
        PIIMiddleware(
            "email",
            strategy="mask"
        ),
    ],
)

response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": """
My email is john.smith@gmail.com.
Please acknowledge my email address.
"""
        }
    ]
})

print(response["messages"][-1].content)
```

```python
#Strategy: Block

from langchain.agents import create_agent
from langchain.agents.middleware import PIIMiddleware

agent = create_agent(
   model=llm_model,
    middleware=[
        PIIMiddleware(
            "api_key",
            detector=r"sk-[a-zA-Z0-9]{32}",
            strategy="block"
        ),
    ],
)


response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "My api key is sk-2345rr"
        }
    ]
})

print(response["messages"][-1].content)
```

```python
### CUSTOM PII -- Aadhaar example

from langchain.agents import create_agent
from langchain.agents.middleware import PIIMiddleware

aadhaar_detector = r"\b\d{4}\s?\d{4}\s?\d{4}\b"

agent = create_agent(
    model = llm_model,
    middleware=[
        PIIMiddleware(
            "aadhaar",
            detector=aadhaar_detector,
            strategy="mask"
        ),
    ],
)

response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "My adhaarr number is 233 -445  -4455"
        }
    ]
})

print(response["messages"][-1].content)
```
